## Part 5: Principal Components Analysis

In this question, we will perform a principal components on a dataset of four morphological measurements taken from individuals representing three species of penguins. We will import the data file in the cell below, but you may wish to create a new cell to examine structure of the dataset as you run the provided code and answer the following questions.

Please run the code in the cell below. Do not modify it.

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

dat=pd.read_csv("penguins.csv")
dat=dat.dropna(axis=0)
traits=dat.loc[:,dat.columns[2:-1]].values
traits=pd.DataFrame(StandardScaler().fit_transform(traits))

pca=PCA(n_components=4) # set up our PCA 
fit=pca.fit_transform(traits) # fit the PCA to our scaled measurements
scores = pd.DataFrame(data = fit, columns = ['pc1', 'pc2', 'pc3', 'pc4'])
```

### Task 1. Interpreting PCA results

**Q5-1a)** Please run the code cell below. What do the numbers tell us about the variance across the four measurements in our dataset? Explain the concept of "redundant" information and what the values below tell us about how much redundancy there is across the variables. _(4pts)_

```python
## Run this cell. Do not modify it.

prop_var=pca.explained_variance_ratio_ * 100
prop_var
```

**WRITE ANSWER HERE**

**Q5-1b)** Please run the code cell below. What does the plot tell us about the relationships between our four variables? Are any of them positively or negatively correlated? _(2pts)_

```python
## Run this cell. Do not modify it.

ax=plt.plot(scores.pc1, scores.pc2, marker='o', linestyle='')
coeff = np.transpose(pca.components_[0:3,:])
n = coeff.shape[0]
for i in range(n):
    plt.arrow(0, 0, coeff[i,0]*60, coeff[i,1]*60,color = 'r',alpha = 0.9)
    plt.text(coeff[i,0]* 60, coeff[i,1] * 60, list(dat.columns[2:])[i], color = 'g', ha = 'center', va = 'center')
plt.xlabel("PC1 ("+str(round(prop_var[0]))+"%)")
plt.ylabel("PC2 ("+str(round(prop_var[1]))+"%)")
plt.show()
```
